```python
import cv2 as cv
import os
```

```python
folder = r"E:\Work\SWIFTT-BTI-2\training\media"

A = [(173, 233), (357, 378)]
B = [(550, 240), (724, 384)]
C = [(578, 602), (765, 733)]

points = [A, B, C]
colors = [[255, 0, 0], [0, 255, 0], [0, 0, 255]]
s = 9

for f in os.listdir(folder):
    if not f.endswith(".png"):
        continue

    img = cv.imread(os.path.join(folder, f))

    for p, color in zip(points, colors):
        img = cv.rectangle(img, (p[0][0] - s // 2, p[0][1] - s // 2), (p[1][0] + s // 2, p[1][1] + s // 2), [0, 0, 0], s)
        img = cv.rectangle(img, (p[0][0] + s // 2, p[0][1] + s // 2), (p[1][0] - s // 2, p[1][1] - s // 2), [0, 0, 0], s)
        img = cv.rectangle(img, p[0], p[1], color, s)

    cv.imwrite(os.path.join(folder, "rect", f), img)
```
