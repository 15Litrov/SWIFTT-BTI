```python
from osgeo import gdal
import helper
import os
import numpy as np
```

```python
stressed = helper.getStressedImagesNames('stress_date.xlsx')
print(len(stressed))
```

```python
stressed
```

```python
coniferous_count = 0
stressed_count = 0

for (n, _) in stressed:
    ft_ds = gdal.Open(os.path.join("SWIFTT_dataset_France", str(n), f"geojson_{n}_forest_type_2018.tif"))
    ft = ft_ds.GetRasterBand(1).ReadAsArray()
    u = np.unique(ft, return_counts=True)

    coniferous_count += u[1][np.nonzero(u[0] == 2)[0]]

    gt_ds = gdal.Open(os.path.join("SWIFTT_dataset_France", str(n), f"geojson_{n}_mask.tif"))
    gt = gt_ds.GetRasterBand(1).ReadAsArray()
    u = np.unique(gt, return_counts=True)
    stressed_count += u[1][np.nonzero(u[0] == 1)[0]]

print(coniferous_count, stressed_count)
```
