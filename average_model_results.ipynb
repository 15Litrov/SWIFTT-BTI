```python
import pandas as pd
import os
```

```python
results_root = r"training\results"

tables = {}

for i in range(3):
    folder = os.path.join(results_root, f"run_{i}")
    for f in os.listdir(folder):
        if not f.endswith(".csv"):
            continue

        table_key = f.split('-')[-3]

        if f not in tables:
            tables[table_key] = []

        tables[table_key].append(pd.read_csv(os.path.join(folder, f)))

for t in tables.keys():
    tables[t] = pd.concat(tables[t])
    tables[t] = tables[t].groupby("mode").mean()
```

```python
for table in tables.keys():
    tables[table].to_csv(f"results_mean\{table}.csv")
```
