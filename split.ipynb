```python
import os
import pandas as pd
import numpy as np
```

```python
groups = [f.split('_')[0] for f in os.listdir("dataset_cropped_64_coniferous")]
unique_groups = set(groups)
```

```python
group_dst = {}
for group in unique_groups:
    p = np.random.uniform()
    if p < 0.66:
        split = "TRAIN"
    elif p < 0.83:
        split = "VALID"
    else:
        split = "TEST"

    group_dst[group] = split
```

```python
dst = {
    "image_name": [], "image_path": [], "mask_path": [], "split_name": [], "dir_name": []
}

counter_dict = {g: 0 for g in ["TRAIN", "TEST", "VALID"]}

for f in os.listdir("dataset_cropped_64_coniferous"):
    if not f.endswith("_S2.tif"):
        continue

    dst["image_name"].append("_".join(f.split("_")[:-1]))
    dst["image_path"].append("")
    dst["mask_path"].append("")
    dst["split_name"].append(group_dst[f.split("_")[0]])
    dst["dir_name"].append("data")

    counter_dict[group_dst[f.split("_")[0]]] += 1

print(counter_dict)

df = pd.DataFrame(dst)
df.to_csv("split_64_perblock.csv")
```
